Below is some simple pandas code to only select the dutch cities from the data we collected. Including the two belgian cities will significantly increase the time needed to run scraper 2 and therefore we just focus on the netherelands.

```python
import pandas as pd

#read the data
df = pd.read_csv("../data/raw/hyrox_results_2026-11-20_16-17.csv")

#the dutch cities we want to keep, the belgian ones (BRU26 and GNE25) are left out
dutch_cities = ["AMS26", "GRQ26", "ROT26", "MAASTRICHT25", "Utrecht2025"]

#only keep the rows where the city is in the dutch list
df_dutch = df[df["city_of_event"].isin(dutch_cities)]

#save as csv and json
df_dutch.to_csv("../data/raw/hyrox_results_DUTCH_2026-11-20_16-17.csv", index=False, encoding="utf-8")
df_dutch.to_json("../data/raw/hyrox_results_DUTCH_2026-11-20_16-17.json", orient="records", indent=2, force_ascii=False)

print("Rows before:", len(df))
print("Rows after:", len(df_dutch))
```
